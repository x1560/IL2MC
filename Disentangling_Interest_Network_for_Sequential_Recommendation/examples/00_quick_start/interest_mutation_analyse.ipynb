```python
import numpy as np
import matplotlib.pyplot as plt
import ruptures as rpt
```

# 导入时间序列数据
生成一个随机时间序列，用于演示

```python
np.random.seed(0)
time_series = np.random.randn(100)
```

```python
# 使用Pelt方法进行突变点检测
model = "l2" # 使用平方损失函数
algo = rpt.Pelt(model=model).fit(time_series)
result = algo.predict(pen=2) # 指定惩罚项penalty
# 绘制时间序列和突变点
plt.plot(time_series, label="time series")
for i in range(len(result)):
    plt.axvline(result[i], color="red", linestyle="--", label="change point")
plt.legend()
plt.show()
# 返回突变点位置和概率
result_prob = algo.predict_proba(pen=2)
print("突变点位置：", result)
print("突变点概率：", result_prob[result])
```
